```python
import pandas as pd
import numpy as np
import math
from scipy import stats
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
statusfile = '/data/kball/SH3_binding/salt_paper_analysis/ArkA12_binding_NaCl/ArkA12_status.csv'
```

```python
# set number of independent simulations

num_Apo_NaCl_sims = 10
num_ArkA17_bound_NaCl_sims = 10
num_ArkA12_bound_NaCl_sims = 10
num_ArkA12_binding_NaCl_sims = 50

len_Apo_NaCl_sims = 240000
len_ArkA17_bound_NaCl_sims = 90000
len_ArkA12_bound_NaCl_sims = 240000
len_ArkA12_binding_NaCl_sims = 100000
```

```python
# calculate ion atmosphere for apo sims

Apo_Na_ions = []
Apo_Cl_ions = []
Apo_9A_Na_ions = []
Apo_9A_Cl_ions = []

for sim in range(0,num_Apo_NaCl_sims):
    Apo_Na_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_NaCl/3A_Na_sim' + str(sim+1) + '.dat', sep = '\s+')
    Apo_Cl_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_NaCl/3A_Cl_sim' + str(sim+1) + '.dat', sep = '\s+')
    Apo_9A_Na_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_NaCl/9A_Na_sim' + str(sim+1) + '.dat', sep = '\s+')
    Apo_9A_Cl_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_NaCl/9A_Cl_sim' + str(sim+1) + '.dat', sep = '\s+')
    
    Apo_Na_ions.append(len(Apo_Na_ions_sim)/(len_Apo_NaCl_sims))
    Apo_Cl_ions.append(len(Apo_Cl_ions_sim)/(len_Apo_NaCl_sims))
    Apo_9A_Na_ions.append(len(Apo_9A_Na_ions_sim)/(len_Apo_NaCl_sims))
    Apo_9A_Cl_ions.append(len(Apo_9A_Cl_ions_sim)/(len_Apo_NaCl_sims))

Apo_Na_avg = np.mean(Apo_Na_ions)
Apo_Na_sem = stats.sem(Apo_Na_ions)
print("Apo Na+ 3A = " + str(Apo_Na_avg) + " +/- " + str(Apo_Na_sem))

Apo_Cl_avg = np.mean(Apo_Cl_ions)
Apo_Cl_sem = stats.sem(Apo_Cl_ions)
print("Apo Cl- 3A = " + str(Apo_Cl_avg) + " +/- " + str(Apo_Cl_sem))

Apo_9A_Na_avg = np.mean(Apo_9A_Na_ions)
Apo_9A_Na_sem = stats.sem(Apo_9A_Na_ions)
print("Apo Na+ 9A = " + str(Apo_9A_Na_avg) + " +/- " + str(Apo_9A_Na_sem))

Apo_9A_Cl_avg = np.mean(Apo_9A_Cl_ions)
Apo_9A_Cl_sem = stats.sem(Apo_9A_Cl_ions)
print("Apo Cl- 9A = " + str(Apo_9A_Cl_avg) + " +/- " + str(Apo_9A_Cl_sem))

print("Apo net 3A = "+ str(Apo_Na_avg-Apo_Cl_avg) + " +/- " + str(Apo_Na_sem+Apo_Cl_sem) )
```

```python
# calculate ion atmosphere for ArkA17 bound sims

ArkA17_bound_Na_ions = []
ArkA17_bound_Cl_ions = []
ArkA17_bound_9A_Na_ions = []
ArkA17_bound_9A_Cl_ions = []

for sim in range(0,num_ArkA17_bound_NaCl_sims):
    ArkA17_bound_Na_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_ArkA17_NaCl/3A_Na_sim' + str(sim+1) + '.dat', sep = '\s+')
    ArkA17_bound_Cl_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_ArkA17_NaCl/3A_Cl_sim' + str(sim+1) + '.dat', sep = '\s+')
    ArkA17_bound_9A_Na_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_ArkA17_NaCl/9A_Na_sim' + str(sim+1) + '.dat', sep = '\s+')
    ArkA17_bound_9A_Cl_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_ArkA17_NaCl/9A_Cl_sim' + str(sim+1) + '.dat', sep = '\s+')
    
    ArkA17_bound_Na_ions.append(len(ArkA17_bound_Na_ions_sim)/(len_ArkA17_bound_NaCl_sims))
    ArkA17_bound_Cl_ions.append(len(ArkA17_bound_Cl_ions_sim)/(len_ArkA17_bound_NaCl_sims))
    ArkA17_bound_9A_Na_ions.append(len(ArkA17_bound_9A_Na_ions_sim)/(len_ArkA17_bound_NaCl_sims))
    ArkA17_bound_9A_Cl_ions.append(len(ArkA17_bound_9A_Cl_ions_sim)/(len_ArkA17_bound_NaCl_sims))

ArkA17_bound_Na_avg = np.mean(ArkA17_bound_Na_ions)
ArkA17_bound_Na_sem = stats.sem(ArkA17_bound_Na_ions)
print("ArkA17_bound Na+ 3A = " + str(ArkA17_bound_Na_avg) + " +/- " + str(ArkA17_bound_Na_sem))

ArkA17_bound_Cl_avg = np.mean(ArkA17_bound_Cl_ions)
ArkA17_bound_Cl_sem = stats.sem(ArkA17_bound_Cl_ions)
print("ArkA17_bound Cl- 3A = " + str(ArkA17_bound_Cl_avg) + " +/- " + str(ArkA17_bound_Cl_sem))

ArkA17_bound_9A_Na_avg = np.mean(ArkA17_bound_9A_Na_ions)
ArkA17_bound_9A_Na_sem = stats.sem(ArkA17_bound_9A_Na_ions)
print("ArkA17_bound Na+ 9A = " + str(ArkA17_bound_9A_Na_avg) + " +/- " + str(ArkA17_bound_9A_Na_sem))

ArkA17_bound_9A_Cl_avg = np.mean(ArkA17_bound_9A_Cl_ions)
ArkA17_bound_9A_Cl_sem = stats.sem(ArkA17_bound_9A_Cl_ions)
print("ArkA17_bound Cl- 9A = " + str(ArkA17_bound_9A_Cl_avg) + " +/- " + str(ArkA17_bound_9A_Cl_sem))
```

```python
# calculate ion atmosphere for ArkA12 bound sims

ArkA12_bound_Na_ions = []
ArkA12_bound_Cl_ions = []
ArkA12_bound_9A_Na_ions = []
ArkA12_bound_9A_Cl_ions = []

for sim in range(0,num_ArkA12_bound_NaCl_sims):
    ArkA12_bound_Na_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_ArkA12_NaCl/3A_Na_sim' + str(sim+1) + '.dat', sep = '\s+')
    ArkA12_bound_Cl_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_ArkA12_NaCl/3A_Cl_sim' + str(sim+1) + '.dat', sep = '\s+')
    ArkA12_bound_9A_Na_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_ArkA12_NaCl/9A_Na_sim' + str(sim+1) + '.dat', sep = '\s+')
    ArkA12_bound_9A_Cl_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/AbpSH3_ArkA12_NaCl/9A_Cl_sim' + str(sim+1) + '.dat', sep = '\s+')
    
    ArkA12_bound_Na_ions.append(len(ArkA12_bound_Na_ions_sim)/(len_ArkA12_bound_NaCl_sims))
    ArkA12_bound_Cl_ions.append(len(ArkA12_bound_Cl_ions_sim)/(len_ArkA12_bound_NaCl_sims))
    ArkA12_bound_9A_Na_ions.append(len(ArkA12_bound_9A_Na_ions_sim)/(len_ArkA12_bound_NaCl_sims))
    ArkA12_bound_9A_Cl_ions.append(len(ArkA12_bound_9A_Cl_ions_sim)/(len_ArkA12_bound_NaCl_sims))

ArkA12_bound_Na_avg = np.mean(ArkA12_bound_Na_ions)
ArkA12_bound_Na_sem = stats.sem(ArkA12_bound_Na_ions)
print("ArkA12_bound Na+ 3A = " + str(ArkA12_bound_Na_avg) + " +/- " + str(ArkA12_bound_Na_sem))

ArkA12_bound_Cl_avg = np.mean(ArkA12_bound_Cl_ions)
ArkA12_bound_Cl_sem = stats.sem(ArkA12_bound_Cl_ions)
print("ArkA12_bound Cl- 3A = " + str(ArkA12_bound_Cl_avg) + " +/- " + str(ArkA12_bound_Cl_sem))

ArkA12_bound_9A_Na_avg = np.mean(ArkA12_bound_9A_Na_ions)
ArkA12_bound_9A_Na_sem = stats.sem(ArkA12_bound_9A_Na_ions)
print("ArkA12_bound Na+ 9A = " + str(ArkA12_bound_9A_Na_avg) + " +/- " + str(ArkA12_bound_9A_Na_sem))

ArkA12_bound_9A_Cl_avg = np.mean(ArkA12_bound_9A_Cl_ions)
ArkA12_bound_9A_Cl_sem = stats.sem(ArkA12_bound_9A_Cl_ions)
print("ArkA12_bound Cl- 9A = " + str(ArkA12_bound_9A_Cl_avg) + " +/- " + str(ArkA12_bound_9A_Cl_sem))
```

```python
# calculate ion atmosphere for ArkA12 Encounter Complex

if statusfile != "none":
    stat = pd.read_csv(statusfile)

ArkA12_EC_Na_ions = []
ArkA12_EC_Cl_ions = []
ArkA12_EC_9A_Na_ions = []
ArkA12_EC_9A_Cl_ions = []

Encounter_frames = pd.DataFrame(columns = ['EC'])
Encounter_frames['EC'] = (stat['status'] != 'unbound') & (stat['status'] != 'Bound') # tell whether frame is in encounter complex
Encounter_frames.index = Encounter_frames.index + 1
Encounter_frames_dict = Encounter_frames.to_dict()

for sim in range(0,num_ArkA12_binding_NaCl_sims):
    Encounter_frames_num = Encounter_frames['EC'][sim*len_ArkA12_binding_NaCl_sims:(sim+1)*len_ArkA12_binding_NaCl_sims].sum() # number of encounter frames for that simulation
    
    ArkA12_binding_Na_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/ArkA12_binding_NaCl/ion_atmosphere/3A_Na_sim' + str(sim+1) + '.dat', sep = '\s+')
    ArkA12_binding_Cl_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/ArkA12_binding_NaCl/ion_atmosphere/3A_Cl_sim' + str(sim+1) + '.dat', sep = '\s+')
    ArkA12_binding_9A_Na_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/ArkA12_binding_NaCl/ion_atmosphere/9A_Na_sim' + str(sim+1) + '.dat', sep = '\s+')
    ArkA12_binding_9A_Cl_ions_sim = pd.read_table('/data/kball/SH3_binding/salt_paper_analysis/ArkA12_binding_NaCl/ion_atmosphere/9A_Cl_sim' + str(sim+1) + '.dat', sep = '\s+')

    ArkA12_binding_Na_ions_sim['#Frame'] = ArkA12_binding_Na_ions_sim['#Frame'].add(sim*len_ArkA12_binding_NaCl_sims)
    ArkA12_binding_Na_ions_sim['EC'] = ArkA12_binding_Na_ions_sim['#Frame'].map(Encounter_frames_dict['EC'])
    ArkA12_binding_Cl_ions_sim['#Frame'] = ArkA12_binding_Cl_ions_sim['#Frame'].add(sim*len_ArkA12_binding_NaCl_sims)
    ArkA12_binding_Cl_ions_sim['EC'] = ArkA12_binding_Cl_ions_sim['#Frame'].map(Encounter_frames_dict['EC'])
    ArkA12_binding_9A_Na_ions_sim['#Frame'] = ArkA12_binding_9A_Na_ions_sim['#Frame'].add(sim*len_ArkA12_binding_NaCl_sims)
    ArkA12_binding_9A_Na_ions_sim['EC'] = ArkA12_binding_9A_Na_ions_sim['#Frame'].map(Encounter_frames_dict['EC'])
    ArkA12_binding_9A_Cl_ions_sim['#Frame'] = ArkA12_binding_9A_Cl_ions_sim['#Frame'].add(sim*len_ArkA12_binding_NaCl_sims)
    ArkA12_binding_9A_Cl_ions_sim['EC'] = ArkA12_binding_9A_Cl_ions_sim['#Frame'].map(Encounter_frames_dict['EC'])
    
    ArkA12_EC_Na_ions.append(ArkA12_binding_Na_ions_sim['EC'].sum()/(Encounter_frames_num))
    ArkA12_EC_Cl_ions.append(ArkA12_binding_Cl_ions_sim['EC'].sum()/(Encounter_frames_num))
    ArkA12_EC_9A_Na_ions.append(ArkA12_binding_9A_Na_ions_sim['EC'].sum()/(Encounter_frames_num))
    ArkA12_EC_9A_Cl_ions.append(ArkA12_binding_9A_Cl_ions_sim['EC'].sum()/(Encounter_frames_num))

ArkA12_EC_Na_avg = np.mean(ArkA12_EC_Na_ions)
ArkA12_EC_Na_sem = stats.sem(ArkA12_EC_Na_ions)
print("ArkA12_EC Na+ 3A = " + str(ArkA12_EC_Na_avg) + " +/- " + str(ArkA12_EC_Na_sem))

ArkA12_EC_Cl_avg = np.mean(ArkA12_EC_Cl_ions)
ArkA12_EC_Cl_sem = stats.sem(ArkA12_EC_Cl_ions)
print("ArkA12_EC Cl- 3A = " + str(ArkA12_EC_Cl_avg) + " +/- " + str(ArkA12_EC_Cl_sem))

ArkA12_EC_9A_Na_avg = np.mean(ArkA12_EC_9A_Na_ions)
ArkA12_EC_9A_Na_sem = stats.sem(ArkA12_EC_9A_Na_ions)
print("ArkA12_EC Na+ 9A = " + str(ArkA12_EC_9A_Na_avg) + " +/- " + str(ArkA12_EC_9A_Na_sem))

ArkA12_EC_9A_Cl_avg = np.mean(ArkA12_EC_9A_Cl_ions)
ArkA12_EC_9A_Cl_sem = stats.sem(ArkA12_EC_9A_Cl_ions)
print("ArkA12_EC Cl- 9A = " + str(ArkA12_EC_9A_Cl_avg) + " +/- " + str(ArkA12_EC_9A_Cl_sem))
```
