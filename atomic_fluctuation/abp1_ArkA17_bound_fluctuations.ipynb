```python
#comparing the last 300ns to the whole 300ns
bound800['atomnum']= bound800["#Res"]
arka_bound800_x = bound800.loc[58:76, "atomnum"]
# arka_bound800_x = range(6, -11, -1)
arka_bound800_y = bound800.loc[58:76, "mean"]
yerr_arka_bound800 = bound800.loc[58:76, "sem"]

SH3_bound800_x = bound800.loc[1:58, "atomnum"]
SH3_bound800_y = bound800.loc[1:58, "mean"]
yerr_SH3_bound800 = bound800.loc[1:58, "sem"]


plt.plot(arka_bound800_x, arka_bound800_y, color = "green", label = "ArkA17", linewidth = 3 )
plt.fill_between(arka_bound800_x, arka_bound800_y - yerr_arka_bound800, arka_bound800_y + yerr_arka_bound800, facecolor="green", alpha = 0.3)

plt.plot(SH3_bound800_x, SH3_bound800_y, color = "blueviolet", label = "ABP1SH3", linewidth = 3 )
plt.fill_between(SH3_bound800_x, SH3_bound800_y - yerr_SH3_bound800, SH3_bound800_y + yerr_SH3_bound800, facecolor="blueviolet", alpha = 0.3)


plt.xlabel("Residue #", fontsize = 12)
plt.ylabel("Atomic Fluctuation($\AA$)", fontsize = 18)
plt.yticks(size=16)
#plt.xticks(current_x,size=12)
plt.title("800mM NaCl ABP1-ArkA17")
plt.legend(loc = "upper center", fontsize = 12)
#plt.savefig('/data/jcardoso/salt_binding_figures/ArkA17/fluctuation_analysis/salt/plots/ArkA17-SH3_bound800_Atomic_Fluc.png', dpi=2000)
```
