```python
# read in the distance file from the ptraj
# Only 7 distances are being read in because they are the distances calculated from the residues that affect the binding of ArkA and SH3 the most
salt_data_1 = salt_resid9 = pd.read_table(
    '/data/jcardoso/data/binding_figures_800ArkA12/ArkA12/pair_wise_distances/binding/salt/distance_files/9-60.dat',
    sep='\s+')
salt_data_2 = salt_resid54 = pd.read_table(
    '/data/jcardoso/data/binding_figures_800ArkA12/ArkA12/pair_wise_distances/binding/salt/distance_files/54-61.dat',
    sep='\s+')
salt_data_3 = salt_resid53 = pd.read_table(
    '/data/jcardoso/data/binding_figures_800ArkA12/ArkA12/pair_wise_distances/binding/salt/distance_files/53-63.dat',
    sep='\s+')
salt_data_4 = salt_resid37 = pd.read_table(
    '/data/jcardoso/data/binding_figures_800ArkA12/ArkA12/pair_wise_distances/binding/salt/distance_files/37-66.dat',
    sep='\s+')
salt_data_5 = salt_resid35 = pd.read_table(
    '/data/jcardoso/data/binding_figures_800ArkA12/ArkA12/pair_wise_distances/binding/salt/distance_files/35-67.dat',
    sep='\s+')
salt_data_6 = salt_resid33 = pd.read_table(
    '/data/jcardoso/data/binding_figures_800ArkA12/ArkA12/pair_wise_distances/binding/salt/distance_files/33-70.dat',
    sep='\s+')
salt_data_7 = salt_resid15 = pd.read_table(
    '/data/jcardoso/data/binding_figures_800ArkA12/ArkA12/pair_wise_distances/binding/salt/distance_files/15-71.dat',
    sep='\s+')
```

```python
# Combines all the above distance file into a table
# concat = concartinate ( puts all the specified index into a table separated in columns)
# iloc = index location = specifies the rows and columns that you need to concartinate
# : = all = all rows
# 1 = the colums you need from all the individual distance file
# binding_combined = pd.concat([salt_resid9.iloc[:,1],salt_resid54.iloc[:,1],salt_resid53.iloc[:,1],salt_resid37.iloc[:,1],salt_resid35.iloc[:,1],salt_resid33.iloc[:,1],salt_resid15.iloc[:,1]],axis=1)
salt_binding_combined = pd.concat([eval('salt_data_%s.iloc[:,1]' % (i)) for i in range(1, 8)], axis=1)
salt_binding_combined
```

```python
# calculate the average across the rows 
salt_binding_combined['avg']= salt_binding_combined.iloc[:,0:7].mean(axis=1)
salt_binding_combined
```

```python
# Calculate the stdev
salt_binding_combined['stdev']=salt_binding_combined.iloc[:,0:7].std(axis=1)
salt_binding_combined
```

```python
#salt_binding_combined
# add a column called frame
# (,100) counting in 100reds
# np.arange(0,10)*0.1
salt_binding_combined['Time_ns'] = np.arange(0, len(salt_binding_combined)) / 100
```
